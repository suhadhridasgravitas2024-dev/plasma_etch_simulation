```python

import numpy as np
import matplotlib.pyplot as plt
from src.plasma_functions import mean_ev_vs_Z_simulation

# --- User Configurable Inputs ---
GAP_LENGTH = 2.0
N_PARTICLES = 50000
dt = 2e-10
STEPS = 5000

FIELD_ACCEL = 6.08e11      # bulk field acceleration magnitude (sets bulk T ~ 10 eV)
SHEATH_MARGIN = 0.3
E_REFLECT_EV = 5.0

Z_VALUES = np.linspace(0.0, 1.0, 60)   # dense sweep, wall -> center
SLICE_WIDTH = 0.1

# --- Initialize and Run Simulation ---
sim = mean_ev_vs_Z_simulation(
    gap_length=GAP_LENGTH,
    n_particles=N_PARTICLES,
    dt=dt,
    steps=STEPS,
    field_accel=FIELD_ACCEL,
    sheath_margin=SHEATH_MARGIN,
    e_reflect_ev=E_REFLECT_EV
)

sim.run()

# --- Build the 1D profile ---
mean_E = sim.get_mean_energy_profile(Z_VALUES, SLICE_WIDTH)

# --- Plotting ---
plt.figure(figsize=(7, 5), dpi=120)
plt.plot(Z_VALUES, mean_E, color='black', linewidth=2.0, label='Simulated mean energy')
plt.axhline(4.05, linestyle=':', color='red', linewidth=2.0, label='T = 4.05 eV (reference)')

# A "hot" (high-T) population has electrons with a lot of kinetic energy spread over a broad range;
# a "cold" (low-T) population is tightly clustered near low energy.
# 1 eV corresponds to about 11,600 K
# bulk - equal ion and electron density (treat bulk like maxwellian)
# T is a shorthand for how energetic and spread-out the electron velocity distribution is. 
# (Te-bulk) For a population in local thermodynamic-like equilibrium, the electron energies follow a Maxwellian distribution 

plt.xlabel("Z (cm)", fontsize=12)
plt.ylabel("Mean electron energy (eV)", fontsize=12)
plt.xlim(0, 1.0)
plt.ylim(0, 12)

ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.legend(frameon=False, fontsize=10)
plt.tight_layout()
plt.savefig('/mnt/c/Users/semi/Plasma_simulation/plasma_etch_simulation/simple_simulation/data_/trial_library_run_mean_ev.png')
```
